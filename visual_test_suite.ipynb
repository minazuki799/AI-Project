# Visual Test Suite: Rule-Based Expert System for Alzheimer's Decision Support

This notebook executes representative cases, missing-fact scenarios, contradictory evidence checks, and interactive walkthroughs for **Task 5: Testing and Evaluation**.

**Setup:** Ensure this `.ipynb` file is saved in the exact same directory as `engine.py` and `alzheimer.py`.

```python
from engine import InferenceEngine
from alzheimer import questions, answer, contradictions, RuleAlzheimer

def run_test_case(case_name: str, expected_outcome: str, custom_answers: dict):
    """Helper to run a simulated patient case and print formatted outputs."""
    print("=" * 80)
    print(f"TEST CASE: {case_name}")
    print(f"EXPECTED:  {expected_outcome}")
    print("=" * 80)

    test_answers = {k: "no" for k in answer}
    test_answers.update(custom_answers)

    engine = InferenceEngine(
        questions=questions,
        answers=test_answers,
        rule_class=RuleAlzheimer,
        contradictions=contradictions
    )

    engine.forward_chaining()
    return engine
```

### Test Case 1: Representative Case (2 Warning Signs)
**Expected Outcome:** `Possible cognitive decline` with Confidence `0.60` (100% data completeness).

```python
run_test_case(
    case_name="Representative Case 1 (2 Warning Signs)",
    expected_outcome="Possible cognitive decline | Confidence: 0.60",
    custom_answers={
        "memory_loss": "yes",
        "language": "yes"
    }
);
```

### Test Case 5: Missing / Unknown Facts

### Test Case 6: Contradictory Evidence (Interactive Clarification)
**Expected Outcome:** Triggers the inconsistency warning between `memory_loss` (`'no'`) and `misplacing` (`'yes'`). Type `'no'` at the prompt to verify it logs in `unresolved_contradictions`, or `'yes'` to update your answers.

```python
run_test_case(
    case_name="Contradictory Evidence Case (memory_loss='no' + misplacing='yes')",
    expected_outcome="Triggers inconsistency warning and logs conflict if declined",
    custom_answers={
        "memory_loss": "no",
        "misplacing": "yes",
        "language": "yes"
    }
);
```
